```python
#初始化链表
class Node():
    def __init__(self, value):
        self.value = value
        self.next = None
    
    def __repr__(self):
        """返回结点的字符串表示"""
        return f"Node({self.value!r})"

class LList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None
    
    def append(self, value):
        new_node = Node(value)
        if self.is_empty():
            self.head = new_node
            return
        current = self.head
        while current.next is not None:
            current = current.next
        current.next = new_node

    def __repr__(self):
        nodes = []
        current = self.head
        while current is not None:
            nodes.append(repr(current.value))
            current = current.next
        return " -> ".join(nodes)
```

作业 1：实现链表反转 reverse()

```python
@classmethod
def reverse(cls, head: Node) -> Node:
    pre = None
    cur = head
    while cur:
        next_node = cur.next  # 保存下一个节点，防止断链
        cur.next = pre        # 反转当前节点的指向
        pre = cur             # pre 后移
        cur = next_node       # cur 后移
    return pre  # 循环结束时 pre 就是新的头节点

# 绑定到类
LList.reverse = reverse

# 构建链表 a -> b -> c
mlist1 = LList()
for i in ["a", "b", "c"]:
    mlist1.append(i)

print("原始链表:", mlist1)

# 反转
new_head = LList.reverse(mlist1.head)

# 关键：把链表的头更新为反转后的新头
mlist1.head = new_head

# 现在打印就是正确的了
print("反转后的链表:", mlist1)
```

作业 2：检测链表是否有环 has_cycle()

核心思路：快慢指针法（Floyd 判圈算法）
慢指针每次走 1 步，快指针每次走 2 步
如果链表无环，快指针会先走到 None
如果链表有环，快慢指针一定会在环内相遇

```python
@classmethod
def has_cycle(cls, head):
    if not head:
        return False
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False

LList.has_cycle = has_cycle

# 测试无环
mlist2 = LList()
mlist2.append(1)
mlist2.append(2)
mlist2.append(3)
print(mlist2.has_cycle(mlist2.head))  # False

# 制造环
tail = mlist2.head
while tail.next:
    tail = tail.next
tail.next = mlist2.head.next

print(mlist2.has_cycle(mlist2.head))  # True
```
