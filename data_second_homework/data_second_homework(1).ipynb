```python
from collections.abc import Sequence
from typing import Any,Iterable
class CustomSet:
    """
    创建一个类，能够实现集合的查询、插入、删除功能
    """
    def __init__(self,iterable:Sequence[Any]):
        """
        :param self: 初始化的集合
        """
        self.iterable=list(iterable)
    
    def find_index(self,value:Any) -> int :
        """
        查询
        :param value: 要查询的元素
        :return: 找到返回索引，未找到返回-1
        """
        l=len(self.iterable)
        for i in range(l):
            if self.iterable[i]==value:
                return i
        return -1

    def insert(self, element: Any) -> bool:
        """
        插入
        :param element:要插入的元素:
        return:插入是否成功
        """
        if self.find_index(element) == -1:
            self.iterable.append(element)
            return True
        return False
    def delete(self,element:Any) -> bool:
        """
        删除
        :param element: 要删除的元素
        :return: 删除成功返回True，未删除返回False
        """
        index=self.find_index(element)
        if index != -1:
            self.iterable.pop(index)
            return True
        return False
    def __str__(self):
        """简洁展示集合内容"""
        return str(self.iterable)
    
    
```

```python
"""应用举例"""

my_set=CustomSet([1,4,5,7,8])

"""插入元素"""
my_set.insert(10)
print("插入10后集合：",my_set)
my_set.insert(20)
print("插入20后集合：",my_set)
my_set.insert(20)
print("再次插入20后集合：",my_set)

"""查找元素"""
print("查找元素5：",my_set.find_index(5))
print("查找元素20：",my_set.find_index(20))
print("查找元素30：",my_set.find_index(30))

"""删除元素"""
my_set.delete(20)
print("删除20后集合：",my_set)
my_set.delete(20)
print("再次删除20后集合（无变化）：",my_set)
```
