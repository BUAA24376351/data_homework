```python
from typing import Any
from collections.abc import Sequence
class OrderedArrayInsert:
    """
    升序有序数组的插入
    """
    def __init__(self,iterable: Sequence[Any]):
        """
        :param iterable: 有序数组初始化
        """
        self.iterable = list(iterable)
        self.length = len(iterable)
    def insert(self,element:Any) -> None:
        """
        插入元素
        :param element: 待插入的元素
        """
        insert_index:int=self.length
        for index,value in enumerate(self.iterable):
            if element < value:
                insert_index=index
                break
        self.iterable.append("")
        for i in range(self.length,insert_index,-1):
            self.iterable[i]=self.iterable[i-1]
        self.iterable[insert_index]=element
        self.length+=1
        
    def __str__(self):
        """打印对象时直接显示有序数组"""
        return str(self.iterable)
```

```python
array=[1,4,21,59]
base_array=OrderedArrayInsert(array)
print("原有序数组：",base_array)
base_array.insert(19)
print("插入19：",base_array)
base_array.insert(5)
print("插入5：",base_array)
base_array.insert(-3)
print("插入-3：",base_array)
base_array.insert(69)
print("插入69：",base_array)
```
