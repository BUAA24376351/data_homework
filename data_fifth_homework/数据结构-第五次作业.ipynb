```python
# 定义链表的节点类
class ListNode:
    def __init__(self, val):
        self.val = val  # 节点存储的值
        self.next = None  # 节点的后继指针，默认指向None

    # 重写__repr__，方便打印查看节点（之前讲的知识点～）
    def __repr__(self):
        return f"Node({self.val})"

# 定义单向链表类
class LinkedList:
    def __init__(self):
        self.head = None  # 链表头节点，默认空

    # 尾部添加节点（方便快速构建测试链表）
    def append(self, val):
        new_node = ListNode(val)
        if not self.head:  # 空链表，头节点直接指向新节点
            self.head = new_node
            return
        # 非空链表，遍历到尾节点
        cur = self.head
        while cur.next:
            cur = cur.next
        cur.next = new_node

    # 核心方法：删除指定索引的节点（索引从0开始）
    def delete_by_index(self, index):
        # 边界情况1：空链表，直接返回
        if not self.head:
            print("链表为空，无法删除")
            return
        # 边界情况2：删除头节点（索引0，无前驱节点）
        if index == 0:
            self.head = self.head.next  # 头节点指向原头节点的后继
            return
        # 步骤1：找到待删除节点的前驱节点（索引index-1）
        cur = self.head  # 从头部开始遍历
        count = 0
        # 遍历条件：cur不是空 + 还没到目标前驱节点
        while cur and count < index - 1:
            cur = cur.next
            count += 1
        # 边界情况3：索引越界（前驱节点为空 或 前驱节点无后继）
        if not cur or not cur.next:
            print(f"索引{index}越界，无法删除")
            return
        # 步骤2：让前驱节点的next指向待删除节点的后继（跳过待删除节点）
        # cur 是前驱节点（比如索引1），cur.next 是待删除节点（比如索引2）
        cur.next = cur.next.next
        # 步骤3：待删除节点因无引用，会被Python自动垃圾回收，无需手动操作

    # 打印链表（方便查看操作结果）
    def print_list(self):
        res = []
        cur = self.head
        while cur:
            res.append(str(cur.val))
            cur = cur.next
        print(" -> ".join(res) if res else "空链表")

# ------------------- 测试案例 -------------------
if __name__ == "__main__":
    # 构建链表：a(0) -> b(1) -> c(2) -> d(3)
    ll = LinkedList()
    for val in ["a", "b", "c", "d"]:
        ll.append(val)
    print("删除前的链表：")
    ll.print_list()  # 输出：a -> b -> c -> d

    # 作业要求：删除索引2的节点（"c"）
    ll.delete_by_index(2)
    print("\n删除索引2后的链表：")
    ll.print_list()  # 输出：a -> b -> d
```
