```python
from collections.abc import Sequence
import random
from typing import Tuple, List


class OptimizedQuickSort:
    """
    优化的快速排序类，包含随机选择轴和三路快速排序。
    """

    def __init__(self, arr: Sequence[int | float]):
        """
        初始化快速排序器。

        :param arr: 待排序的数组，可以是整数或浮点数
        """
        self.arr: list[int | float] = list(arr)
        self.length: int = len(arr)

    def random_partition(self, left_index: int, right_index: int) -> int:
        """
        使用随机轴对数组的指定范围进行分区操作。
        
        随机选择一个轴，将数组分为两部分：
        - 左边的元素都小于轴
        - 右边的元素都大于轴

        :param left_index: 分区范围的左边界索引
        :param right_index: 分区范围的右边界索引
        :return: 轴的最终位置索引
        """
        # 随机选择一个轴
        pivot_index = random.randint(left_index, right_index)
        pivot = self.arr[pivot_index]
        
        # 将轴移到最右边
        self.arr[pivot_index], self.arr[right_index] = (
            self.arr[right_index],
            self.arr[pivot_index],
        )
        
        # 右指针从轴的左边开始
        right_pointer: int = right_index - 1
        left_pointer: int = left_index

        while True:
            # 左指针向右移动，直到找到大于或等于轴的值
            while left_pointer <= right_pointer and self.arr[left_pointer] < pivot:
                left_pointer += 1

            # 右指针向左移动，直到找到小于或等于轴的值
            while left_pointer <= right_pointer and self.arr[right_pointer] > pivot:
                right_pointer -= 1

            # 如果左右指针相遇或交叉，停止
            if left_pointer >= right_pointer:
                break

            # 交换左右指针指向的值
            self.arr[left_pointer], self.arr[right_pointer] = (
                self.arr[right_pointer],
                self.arr[left_pointer],
            )

        # 最后将左指针的值与轴交换
        self.arr[left_pointer], self.arr[right_index] = (
            self.arr[right_index],
            self.arr[left_pointer],
        )

        # 返回轴的最终位置
        return left_pointer

    def three_way_partition(self, left_index: int, right_index: int) -> Tuple[int, int]:
        """
        三路分区：将数组分为三部分：小于轴、等于轴、大于轴。
        
        返回等于轴区域的左右边界索引。

        :param left_index: 分区范围的左边界索引
        :param right_index: 分区范围的右边界索引
        :return: 等于轴区域的左右边界索引 (left_eq, right_eq)
        """
        # 随机选择轴
        pivot_index = random.randint(left_index, right_index)
        pivot = self.arr[pivot_index]
        
        # 将轴交换到最左边
        self.arr[left_index], self.arr[pivot_index] = (
            self.arr[pivot_index],
            self.arr[left_index],
        )
        
        # 初始化指针
        lt = left_index  # 小于轴区域的右边界
        gt = right_index  # 大于轴区域的左边界
        i = left_index + 1  # 当前检查的位置
        
        while i <= gt:
            if self.arr[i] < pivot:
                # 当前元素小于轴，交换到左边
                self.arr[i], self.arr[lt] = self.arr[lt], self.arr[i]
                lt += 1
                i += 1
            elif self.arr[i] > pivot:
                # 当前元素大于轴，交换到右边
                self.arr[i], self.arr[gt] = self.arr[gt], self.arr[i]
                gt -= 1
                # 注意：这里i不增加，因为交换过来的元素需要检查
            else:
                # 当前元素等于轴，留在中间
                i += 1
        
        # lt是第一个等于轴的位置，gt是最后一个等于轴的位置
        return lt, gt

    def sort_with_random_pivot(
        self, left_index: int | None = None, right_index: int | None = None
    ) -> List[int | float]:
        """
        使用随机选择轴的快速排序。

        :param left_index: 排序范围的左边界索引，默认为数组开头
        :param right_index: 排序范围的右边界索引，默认为数组结尾
        :return: 排序后的数组
        """
        # 初始化边界
        if left_index is None:
            left_index = 0
        if right_index is None:
            right_index = self.length - 1

        # 基准情形：如果子数组长度为0或1，不需要排序
        if left_index >= right_index:
            return self.arr

        # 对当前范围进行分区，获取轴的位置
        pivot_position: int = self.random_partition(left_index, right_index)

        # 对轴左边的子数组递归排序
        self.sort_with_random_pivot(left_index, pivot_position - 1)

        # 对轴右边的子数组递归排序
        self.sort_with_random_pivot(pivot_position + 1, right_index)

        return self.arr

    def three_way_sort(
        self, left_index: int | None = None, right_index: int | None = None
    ) -> List[int | float]:
        """
        三路快速排序，适用于有大量重复元素的数组。

        :param left_index: 排序范围的左边界索引，默认为数组开头
        :param right_index: 排序范围的右边界索引，默认为数组结尾
        :return: 排序后的数组
        """
        # 初始化边界
        if left_index is None:
            left_index = 0
        if right_index is None:
            right_index = self.length - 1

        # 基准情形：如果子数组长度为0或1，不需要排序
        if left_index >= right_index:
            return self.arr

        # 进行三路分区
        left_eq, right_eq = self.three_way_partition(left_index, right_index)

        # 对小于轴的部分递归排序
        self.three_way_sort(left_index, left_eq - 1)

        # 对大于轴的部分递归排序
        self.three_way_sort(right_eq + 1, right_index)

        return self.arr


# 测试代码
if __name__ == "__main__":
    print("=" * 50)
    print("测试随机选择轴的快速排序")
    print("=" * 50)
    
    # 测试随机选择轴
    test_arr1: list[int] = [0, 5, 2, 1, 6, 3]
    print(f"原始数组: {test_arr1}")
    
    qs_random = OptimizedQuickSort(test_arr1.copy())
    sorted_arr1 = qs_random.sort_with_random_pivot()
    print(f"随机轴快速排序后数组: {sorted_arr1}")
    print(f"是否排序正确: {sorted_arr1 == sorted(test_arr1)}")
    
    print("\n" + "=" * 50)
    print("测试三路快速排序（针对有重复元素的数组）")
    print("=" * 50)
    
    # 测试三路快速排序
    test_arr2: list[int] = [3, 1, 2, 3, 5, 2, 2, 3, 1, 4]
    print(f"原始数组（有重复元素）: {test_arr2}")
    
    qs_three_way = OptimizedQuickSort(test_arr2.copy())
    sorted_arr2 = qs_three_way.three_way_sort()
    print(f"三路快速排序后数组: {sorted_arr2}")
    print(f"是否排序正确: {sorted_arr2 == sorted(test_arr2)}")
    
    print("\n" + "=" * 50)
    print("性能对比测试（小规模）")
    print("=" * 50)
    
    # 创建包含大量重复元素的数组
    import time
    
    # 测试数组1：随机数组
    random_arr = [random.randint(0, 100) for _ in range(20)]
    
    # 测试数组2：大量重复元素的数组
    repeat_arr = [random.randint(0, 10) for _ in range(20)]
    
    print("\n1. 随机轴快速排序:")
    qs_random_test = OptimizedQuickSort(random_arr.copy())
    start = time.time()
    result_random = qs_random_test.sort_with_random_pivot()
    end = time.time()
    print(f"  随机数组排序时间: {(end - start) * 1000:.2f}毫秒")
    print(f"  排序正确性: {result_random == sorted(random_arr)}")
    
    print("\n2. 三路快速排序:")
    
    # 测试在随机数组上的性能
    qs_three_way_random = OptimizedQuickSort(random_arr.copy())
    start = time.time()
    result_three_way_random = qs_three_way_random.three_way_sort()
    end = time.time()
    print(f"  随机数组排序时间: {(end - start) * 1000:.2f}毫秒")
    print(f"  排序正确性: {result_three_way_random == sorted(random_arr)}")
    
    # 测试在大量重复元素数组上的性能
    qs_three_way_repeat = OptimizedQuickSort(repeat_arr.copy())
    start = time.time()
    result_three_way_repeat = qs_three_way_repeat.three_way_sort()
    end = time.time()
    print(f"  重复元素数组排序时间: {(end - start) * 1000:.2f}毫秒")
    print(f"  排序正确性: {result_three_way_repeat == sorted(repeat_arr)}")
```

性能优势

随机选择轴：

避免最坏情况时间复杂度 O(n²)

期望时间复杂度为 O(n log n)

对已排序或接近排序的数组特别有效

三路快速排序：

对包含大量重复元素的数组，性能显著优于传统快速排序

时间复杂度为 O(n log n)，但在有大量重复元素时接近 O(n)

避免了重复元素的重复比较和交换

使用建议

对于一般随机数组，使用随机选择轴的快速排序

对于已知包含大量重复元素的数组，使用三路快速排序

可以通过分析数组特性动态选择排序策略
